```python
df_clean["sex"] = df_clean["sex"].replace(".", pd.NA)
```

```python
df_clean.loc[
    (df_clean["flipper_length_mm"] < 150) |
    (df_clean["flipper_length_mm"] > 250),
    "flipper_length_mm"
] = pd.NA
```

```python
numeric_columns = [
    "culmen_length_mm",
    "culmen_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]
```

```python
for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="culmen_length_mm",
    y="flipper_length_mm"
)

plt.title("Penguin Beak Length vs Flipper Length")
plt.xlabel("Beak Length (mm)")
plt.ylabel("Flipper Length (mm)")

plt.show()
```

```python
features = [
    "culmen_length_mm",
    "culmen_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

```

```python
X = df_clean[features]
```

```python
print(X.shape)
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaling tool
scaler = StandardScaler()

# Scale the selected features
X_scaled = scaler.fit_transform(X)

print("Original data:")
display(X.head())

print("\nScaled data:")
print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans

# Create the K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train the model and assign each penguin to a group
clusters = kmeans.fit_predict(X_scaled)

# Add the group number to our dataset
df_clean["Cluster"] = clusters

print("K-Means model trained successfully!")

print("\nFirst 10 penguins and their groups:")
print(df_clean[["Cluster"]].head(10))
```

```python
print("Number of penguins in each cluster:")

print(df_clean["Cluster"].value_counts().sort_index())
```

```python
from sklearn.cluster import KMeans

# Create the K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train the model and assign each penguin to a group
clusters = kmeans.fit_predict(X_scaled)

# Add the group number to our dataset
df_clean["Cluster"] = clusters

print("K-Means model trained successfully!")

print("\nFirst 10 penguins and their groups:")
print(df_clean[["Cluster"]].head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="culmen_length_mm",
    y="flipper_length_mm",
    hue="Cluster",  #gives diff color
    palette="Set1"
)

plt.title("Penguin Clusters Using K-Means")
plt.xlabel("Beak Length (mm)")
plt.ylabel("Flipper Length (mm)")
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, clusters)

print("Silhouette Score:", round(score, 3))
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)

print("WCSS values:")
print(wcss)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    wcss,
    marker="o"
)

plt.title("Elbow Method for Penguin Clustering")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.xticks(range(1, 11))
plt.grid(True)

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 7):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

```python
from sklearn.cluster import KMeans

# Create the final model
final_model = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

# Assign each penguin to a cluster
final_clusters = final_model.fit_predict(X_scaled)

# Save the cluster labels
df_clean["Cluster"] = final_clusters

print("Final K-Means model trained successfully!")

print("\nNumber of penguins in each cluster:")
print(df_clean["Cluster"].value_counts().sort_index())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="culmen_length_mm",
    y="flipper_length_mm",
    hue="Cluster",
    palette="Set1"
)

plt.title("Final Penguin Clusters - K-Means")
plt.xlabel("Beak Length (mm)")
plt.ylabel("Flipper Length (mm)")
plt.legend(title="Cluster")
plt.show()
```

```python
# Save the final dataset to a CSV file
df_clean.to_csv("/content/penguin_clusters.csv", index=False)

print("Dataset saved successfully!")
print("File name: penguin_clusters.csv")
```
